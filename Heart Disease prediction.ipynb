```python
y_pred_final = final_pipeline.predict(X_test_clean)

print("Final Accuracy:", accuracy_score(y_test_clean, y_pred_final))
print("Final Precision:", precision_score(y_test_clean, y_pred_final))
print("Final Recall:", recall_score(y_test_clean, y_pred_final))
print("Final F1:", f1_score(y_test_clean, y_pred_final))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clean, y_pred_final))
```
